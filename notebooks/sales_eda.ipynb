```python
q3 =df_clean["SALES"].quantile(0.75)
q1 =df_clean["SALES"].quantile(0.25)
iqr =q3-q1
lower_limit =q1 -1.5 *iqr
upper_limit =q3  +1.5 *iqr
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
outliers =df_clean[(df_clean["SALES"] < lower_limit) | (df_clean["SALES"] > upper_limit)]
print("Total potential outliers:", len(outliers))
```

```python
outliers[["ORDER_NUMBER", "SALES", "QUANTITY_ORDERED", "PRODUCT_LINE"]].head(10)
```

```python
print("Total potential outliers:", len(outliers)) 
```

```python
outlier_perc =len(outliers)/len(df_clean) * 100
print("Percentage of potential outliers:", outlier_perc, "%")
```

The IQR method identified approximately 2.87% of sales records as potential outliers. These records exceed the upper IQR limit of 7,964.86. Since the high sales values may represent legitimate large transactions, they are retained for further analysis.

The IQR method identified 8 potential outliers (approximately 0.28% of records) in the quantity ordered variable. These observations represent unusually high order quantities compared with the central distribution. Since they may correspond to genuine bulk orders, they are retained for further analysis.
